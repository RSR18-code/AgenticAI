```python
import matplotlib.pyplot as plt  # type: ignore
```

```python
sales_2012 = dr[dr["Year"] == 2012].copy()
sales_2012 = sales_2012.sort_values("Sales_Rep_ID")

plt.figure(figsize=(10, 5))
plt.plot(sales_2012["Sales_Rep_ID"], sales_2012["Value"], marker="o")
plt.xlabel("Sales Rep ID")
plt.ylabel("Value")
plt.title("Sales for Year 2012")
plt.grid(True)
plt.show()
```
